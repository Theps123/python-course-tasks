```python
def minimum(arr):
    return min(arr)

def maximum(arr):
    return max(arr)
```

```python
def name_value(my_list):
    return [
        sum(ord(ch) - 96 for ch in word if ch != ' ') * i
        for i, word in enumerate(my_list, start=1)
    ]
```

```python
class List:
    def remove_(self, integer_list, values_list):
        to_remove = set(values_list)
        return [x for x in integer_list if x not in to_remove]
```

```python
def square(n):
    return n ** 2
```

```python
def past(h, m, s):
    return (h * 3600 + m * 60 + s) * 1000
```

```python
def find_average(nums):
    return sum(nums) / len(nums) if nums else 0
```

```python
import math

def is_square(n):
    return n >= 0 and math.isqrt(n) ** 2 == n
```

```python
def opposite(number):
    return -number
```

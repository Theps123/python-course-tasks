```python
from abc import ABC, abstractmethod
import math


class Shape(ABC):
    def __init__(self, name: str):
        self.name = name

    @abstractmethod
    def area(self) -> float:
        """Вычисление площади фигуры."""
        pass

    @abstractmethod
    def show_params(self) -> str:
        """Вывод параметров, необходимых для вычисления площади."""
        pass

    def __str__(self) -> str:
        return self.name

    def __eq__(self, other) -> bool:
        if not isinstance(other, Shape):
            return NotImplemented
        return math.isclose(self.area(), other.area(), rel_tol=1e-9)

    def __lt__(self, other) -> bool:
        if not isinstance(other, Shape):
            return NotImplemented
        return not (self == other) and self.area() < other.area()

    def __gt__(self, other) -> bool:
        if not isinstance(other, Shape):
            return NotImplemented
        return not (self == other) and self.area() > other.area()

    def compare_area(self, other: "Shape") -> str:
        if not isinstance(other, Shape):
            raise TypeError("Сравнивать можно только с другой фигурой (Shape).")
        if self == other:
            return f"Площадь фигуры '{self}' равна площади фигуры '{other}' ({self.area():.2f})"
        elif self > other:
            return f"Площадь фигуры '{self}' ({self.area():.2f}) больше, чем у '{other}' ({other.area():.2f})"
        else:
            return f"Площадь фигуры '{self}' ({self.area():.2f}) меньше, чем у '{other}' ({other.area():.2f})"


class Circle(Shape):
    def __init__(self, radius: float):
        if radius <= 0:
            raise ValueError("Радиус должен быть положительным числом.")
        super().__init__("Круг")
        self.radius = radius

    def area(self) -> float:
        return math.pi * (self.radius ** 2)

    def show_params(self) -> str:
        return f"Параметры фигуры '{self}': радиус = {self.radius}"


class Rectangle(Shape):
    def __init__(self, width: float, height: float):
        if width <= 0 or height <= 0:
            raise ValueError("Ширина и высота должны быть положительными числами.")
        super().__init__("Прямоугольник")
        self.width = width
        self.height = height

    def area(self) -> float:
        return self.width * self.height

    def show_params(self) -> str:
        return f"Параметры фигуры '{self}': ширина = {self.width}, высота = {self.height}"


# Пример использования
if __name__ == "__main__":
    circle = Circle(5)
    rect1 = Rectangle(4, 10)
    rect2 = Rectangle(5, 8)

    print(circle)                # Круг
    print(circle.show_params())
    print(f"Площадь: {circle.area():.2f}\n")

    print(rect1)                 # Прямоугольник
    print(rect1.show_params())
    print(f"Площадь: {rect1.area():.2f}\n")

    # Сравнение через магические методы и текстовый метод
    print("rect1 == rect2:", rect1 == rect2)   # True (40 == 40)
    print("circle > rect1:", circle > rect1)   # True (78.54 > 40)
    print("rect1 < circle:", rect1 < circle)   # True
    print(circle.compare_area(rect1))
    print(rect1.compare_area(rect2))
```

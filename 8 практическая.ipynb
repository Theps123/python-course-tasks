```python
%%writefile project/app/__init__.py
from .calculator import calculate

__all__ = ["calculate"]
```

```python
%%writefile project/app/calculator.py
def calculate(operation: str, value_1: int, value_2: int):
    """Выполняет арифметическую операцию над двумя числами."""
    if operation == "+":
        return value_1 + value_2
    elif operation == "-":
        return value_1 - value_2
    elif operation == "*":
        return value_1 * value_2
    elif operation == "/":
        if value_2 == 0:
            raise ZeroDivisionError("Деление на ноль невозможно.")
        return value_1 / value_2
    elif operation == "**":
        return value_1 ** value_2
    else:
        raise ValueError(
            f"Неизвестная операция: '{operation}'. Доступны: +, -, *, /, **"
        )
```

```python
%%writefile project/tests/test_calculator.py
import os
import sys
import unittest

PROJECT_ROOT = os.path.dirname(os.path.dirname(os.path.abspath(__file__)))
if PROJECT_ROOT not in sys.path:
    sys.path.insert(0, PROJECT_ROOT)

from app.calculator import calculate


class TestCalculator(unittest.TestCase):
    def test_addition(self):
        self.assertEqual(calculate("+", 4, 7), 11)
        self.assertEqual(calculate("+", -3, 3), 0)

    def test_subtraction(self):
        self.assertEqual(calculate("-", 15, 18), -3)
        self.assertEqual(calculate("-", 10, 4), 6)

    def test_multiplication(self):
        self.assertEqual(calculate("*", 5, 5), 25)
        self.assertEqual(calculate("*", -2, 6), -12)

    def test_division(self):
        self.assertEqual(calculate("/", 49, 7), 7)
        self.assertEqual(calculate("/", 5, 2), 2.5)

    def test_division_by_zero(self):
        with self.assertRaises(ZeroDivisionError):
            calculate("/", 10, 0)

    def test_power(self):
        self.assertEqual(calculate("**", 2, 3), 8)
        self.assertEqual(calculate("**", 5, 0), 1)
        self.assertEqual(calculate("**", 2, -1), 0.5)

    def test_invalid_operation(self):
        with self.assertRaises(ValueError):
            calculate("%", 10, 3)


if __name__ == "__main__":
    unittest.main()
```

```python
%%writefile project/main.py
from app.calculator import calculate


def main():
    print("Калькулятор. Доступные операции: +, -, *, /, ** (возведение в степень)")
    operation = input("Введите операцию: ").strip()

    try:
        value_1 = int(input("Введите первое целое число: "))
        value_2 = int(input("Введите второе целое число: "))
        result = calculate(operation, value_1, value_2)
    except ValueError as e:
        print(f"Ошибка ввода: {e}")
    except ZeroDivisionError as e:
        print(f"Математическая ошибка: {e}")
    else:
        print(f"Результат: {value_1} {operation} {value_2} = {result}")


if __name__ == "__main__":
    main()
```

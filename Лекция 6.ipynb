```python
class Cube(object):
    def __init__(self, side=0):
        self.__side = abs(side)

    def get_side(self):
        return self.__side

    def set_side(self, new_side):
        self.__side = abs(new_side)
```

```python
class Person:
    def __init__(self, name):
        self.name = name

    def greet(self, your_name):
        return f"Hello {your_name}, my name is {self.name}"
```

```python
class Guesser:
    def __init__(self, number, lives):
        self.number = number
        self.lives = lives

    def guess(self, n):
        if self.lives <= 0:
            raise ValueError("No lives left")
        if n == self.number:
            return True
        self.lives -= 1
        return False
```

```python
def find_children(santas_list, children):
    return sorted(set(santas_list) & set(children))
```

```python
n = int(input("Введите количество человек (n): "))

if n <= 0:
    print("Количество человек должно быть больше 0.")
else:
    total_age = 0
    for i in range(1, n + 1):
        age = float(input(f"Введите возраст {i}-го человека: "))
        total_age += age

    average_age = total_age / n
    print(f"Средний возраст: {average_age:.2f}")
```

```python
hex_code = input("Введите цветовой код (например, a0351f): ").strip().lstrip('#').lower()

if len(hex_code) != 6 or any(c not in '0123456789abcdef' for c in hex_code):
    print("Ошибка: введите корректный 6-значный шестнадцатеричный код.")
else:
    # Разбиваем код на составляющие R (красный), G (зеленый), B (синий)
    r = int(hex_code[0:2], 16)
    g = int(hex_code[2:4], 16)
    b = int(hex_code[4:6], 16)

    if r == 0 and g == 0 and b == 0:
        print("черный")
    elif r == 255 and g == 255 and b == 255:
        print("белый")
    elif r == g == b:
        print("серый")
    elif r > g and r > b:
        print("Цвет ближе к красному")
    elif g > r and g > b:
        print("Цвет ближе к зеленому")
    elif b > r and b > g:
        print("Цвет ближе к синему")
    else:
        # Случай, когда два максимальных канала равны между собой
        max_val = max(r, g, b)
        colors = []
        if r == max_val:
            colors.append("красному")
        if g == max_val:
            colors.append("зеленому")
        if b == max_val:
            colors.append("синему")
        print("Цвет одинаково близок к:", " и ".join(colors))
```
